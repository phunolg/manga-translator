```python
import json
import os
import requests
from pathlib import Path

BASE_URL = "http://localhost:8000"
STORY_NAME = "Yule"
def test_create_character(story_name, character_data):
    """
    Test API tạo nhân vật mới
    
    Args:
        story_name (str): Tên truyện
        character_data (dict): Thông tin nhân vật
    """
    endpoint = f"{BASE_URL}/init-manga/character"
    
    # Chuẩn bị dữ liệu form
    form_data = {
        "story_name": story_name,
        "name_character": character_data["name"],
        "description": character_data["description"],
        "address_matrix": json.dumps(character_data["address_matrix"])
    }
    
    # Chuẩn bị file hình ảnh
    image_path = character_data["image_path"]
    files = {
        "character_image": (os.path.basename(image_path), open(image_path, "rb"), "image/png")
    }
    
    try:
        # Gọi API
        response = requests.post(endpoint, data=form_data, files=files)
        
        # Đóng file sau khi sử dụng
        files["character_image"][1].close()
        
        # In kết quả
        print(f"Status code: {response.status_code}")
        print(f"Response: {response.json()}")
        
        return response.json()
    except Exception as e:
        print(f"Error: {str(e)}")
        
        # Đảm bảo file được đóng ngay cả khi có lỗi
        if "character_image" in files:
            files["character_image"][1].close()
        
        return {"error": str(e)}

```
